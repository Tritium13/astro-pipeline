```python
from astroquery.mast import Observations
from astropy.time import Time
import numpy as np

now = Time.now().mjd
t = Observations.query_criteria(obs_collection="JWST", instrument_name="NIRCAM/IMAGE", calib_level=3,
                                dataRights="PUBLIC", t_obs_release=[now - 30, now])
t["released"] = [Time(x, format="mjd").iso[:10] for x in t["t_obs_release"]]
t["observed"] = [Time(x, format="mjd").iso[:10] for x in t["t_min"]]
t.sort("t_obs_release", reverse=True)
print(len(t), "products")
t["released", "observed", "target_name", "proposal_id", "filters"].pprint(max_lines=40, max_width=-1)
```

```python
j = Observations.query_criteria(obs_collection="JWST", instrument_name="NIRCAM/IMAGE", calib_level=3,
                                dataRights="PUBLIC", target_name="JUPITER")
j["observed"] = [Time(x, format="mjd").iso[:16] for x in j["t_min"]]
j["obs"] = [o.split("_nircam")[0] for o in j["obs_id"]]
j.sort("t_min")
j["observed", "proposal_id", "obs", "filters", "t_exptime"].pprint(max_lines=-1, max_width=-1)
```
